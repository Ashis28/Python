### Adding Extra Functionalities

not allowing -ve we dont have enough quantity

```python
record = json.loads(js)

print("-----------Menu-----------")
      
for key in record.keys():
    print(key ," ",record[key]["Name"]," ",record[key]["Price"],record[key]["Qn"])

ui_prid = str(input("Enter Product id :"))  #user input pr id
ui_qn = str(input("Enter the Quantity :"))

print("------------Data----------")
if (record[ui_prid]["Qn"]>=int(ui_qn)):
    print("Name :",record[ui_prid]["Name"])
    print("Price :",record[ui_prid]["Price"])
    print("User Qn:",ui_qn)

    #updating the quantity
    record[ui_prid]["Qn"] = record[ui_prid]["Qn"]-int(ui_qn)
    print("------------BIll----------")
    print("Bill iS :",record[ui_prid]["Price"] * int(ui_qn)," RS ")
else:
    print("Sorry we are not having enough quantity")
    print("We are only having :",record[ui_prid]["Qn"] ," ","but user asked ",ui_qn)

    print("Do you want to purchse it y/n :")
    ch = input("Enter your choise ")
    if(ch=='y' or ch=='Y'):
        print("Name :",record[ui_prid]["Name"])
        print("Price :",record[ui_prid]["Price"])
            #updating the quantity
        record[ui_prid]["Qn"] = "0"
        print("------------BIll----------")
        print("Bill iS :",record[ui_prid]["Price"] * record[ui_prid]["Qn"]," RS ")
    else:
        print("Thanks")



js = json.dumps(record)
fd = open("record.json","w")
fd.write(js)
fd.close()
```

Generating Sales Structure

```python
record = json.loads(js)

print("-----------Menu-----------")
      
for key in record.keys():
    print(key ," ",record[key]["Name"]," ",record[key]["Price"],record[key]["Qn"])

#Taking user credentials
ui_username = input("Enter your name: ")
ui_phoneno = input("Enter your phone no.: ")
ui_mail = input("Enter your mail: ")

ui_prid = str(input("Enter Product id :"))  #user input pr id
ui_qn = str(input("Enter the Quantity :"))

print("------------Data----------")
if (record[ui_prid]["Qn"]>=int(ui_qn)):
    print("Name :",record[ui_prid]["Name"])
    print("Price :",record[ui_prid]["Price"])
    print("User Qn:",ui_qn)

    #updating the quantity
    record[ui_prid]["Qn"] = record[ui_prid]["Qn"]-int(ui_qn)
    print("------------BIll----------")
    print("Bill iS :",record[ui_prid]["Price"] * int(ui_qn)," RS ")

    #generating sales to store in txt
    sales = ui_username+","+ui_phoneno+","+ui_mail+","+ui_prid+","+str(record[ui_prid]["Price"] * int(ui_qn))+"\n"
    
else:
    print("Sorry we are not having enough quantity")
    print("We are only having :",record[ui_prid]["Qn"] ," ","but user asked ",ui_qn)

    print("Do you want to purchse it y/n :")
    ch = input("Enter your choise ")
    if(ch=='y' or ch=='Y'):
        print("Name :",record[ui_prid]["Name"])
        print("Price :",record[ui_prid]["Price"])
            #updating the quantity
        record[ui_prid]["Qn"] = "0"
        print("------------BIll----------")
        print("Bill iS :",record[ui_prid]["Price"] * record[ui_prid]["Qn"]," RS ")

        #generating sales to store in txt
        sales = ui_username+","+ui_phoneno+","+ui_mail+","+ui_prid+","+str(record[ui_prid]["Price"] * record[ui_prid]["Qn"])+"\n"
    else:
        print("Thanks")


fd = open("sales.txt","a")
fd.write(sales)
fd.close()

js = json.dumps(record)
fd = open("record.json","w")
fd.write(js)
fd.close()
```
