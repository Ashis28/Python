Adding more functionalities

```python
#Reading the inventory
fd = open("Inventory.txt","r")
products = fd.read().split("\n")
fd.close()

#Taking the inputs
ui_prod_id = input("Enter product id: ") #ui = user input
ui_prod_qt = input("Enter product quantity: ")

updated_list = []

for prod in products:
    prod_details = prod.split(",")
    
    if(prod_details[0]==ui_prod_id):
        
        if(int(prod_details[3]) >= int(ui_prod_qt)):
            
            print("Name :"+prod_details[1])
            print("Price :"+prod_details[2])
            print("Quantity :"+ui_prod_qt)
            print("Bill :",int(prod_details[2])*int(ui_prod_qt))
    
            prod_details[3] = str(int(prod_details[3])-int(ui_prod_qt))
            print("---------------------------")

        else:
            print("Sorry we are not having enough quantiy ")
            print("We are only having",prod_details[3],"Quantity")
            print("Would you like to purchase it Y/N")
            
            ch = input("press Y/N")

            if (ch == 'Y' or ch == 'y'):
                print("Name :"+prod_details[1])
                print("Price :"+prod_details[2])
                print("Quantity :"+ui_prod_qt)
                print("Bill :",int(prod_details[2])*int(ui_prod_qt))

                prod_details[3] = "0"
            else:
                print("Thanks")    
            
    updated_list.append(prod_details)


#saving in txt file
lst = []
fd = open("Inventory.txt","w")
for i in updated_list:
    lst.append((i[0]+","+i[1]+","+i[2]+","+i[3])+"\n")

lst[-1] = lst[-1][:-1] #removing last index from list

for i in lst:
    fd.write(i)
fd.close()
```

Generating Sales

```python
#Noting time transactions
import time

#Reading the inventory
fd = open("Inventory.txt","r")
products = fd.read().split("\n")
fd.close()

#Taking the inputs
ui_prod_id = input("Enter product id: ") #ui = user input
ui_prod_qt = input("Enter product quantity: ")

ui_username = input("Enter your name: ")
ui_phoneno = input("Enter your phone no.: ")
ui_mail = input("Enter your mail: ")

updated_list = []

for prod in products:
    prod_details = prod.split(",")
    
    if(prod_details[0]==ui_prod_id):
        
        if(int(prod_details[3]) >= int(ui_prod_qt)):
            
            print("Name :"+prod_details[1])
            print("Price :"+prod_details[2])
            print("Quantity :"+ui_prod_qt)
            print("Bill :",int(prod_details[2])*int(ui_prod_qt))

            #Inventory Updated
            prod_details[3] = str(int(prod_details[3])-int(ui_prod_qt))

            fd = open("Sales.txt","a")
            sales_detail = ui_username +","+ui_phoneno+","+ui_mail+","+prod_details[1]+","+str(int(prod_details[3])*int(ui_prod_qt))+","+time.ctime()+"\n"
            fd.write(sales_detail)
            fd.close()
            print("---------------------------")

        else:
            print("Sorry we are not having enough quantiy ")
            print("We are only having",prod_details[3],"Quantity")
            print("Would you like to purchase it Y/N")
            
            ch = input("press Y/N")

            if (ch == 'Y' or ch == 'y'):
                print("Name :"+prod_details[1])
                print("Price :"+prod_details[2])
                print("Quantity :"+ui_prod_qt)
                print("Bill :",int(prod_details[2])*int(ui_prod_qt))

                

                fd = open("Sales.txt","a")
                sales_detail = ui_username +","+ui_phoneno+","+ui_mail+","+prod_details[1]+","+str(int(prod_details[3])*int(prod_details[2]))+","+time.ctime()+"\n"
                fd.write(sales_detail)
                fd.close()

                prod_details[3] = "0"
                
            else:
                print("Thanks")    
            
    updated_list.append(prod_details)


#saving in txt file
lst = []
fd = open("Inventory.txt","w")
for i in updated_list:
    lst.append((i[0]+","+i[1]+","+i[2]+","+i[3])+"\n")

lst[-1] = lst[-1][:-1] #removing last index from list

for i in lst:
    fd.write(i)
fd.close()
```
