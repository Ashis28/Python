### File Reading and writing practise from reading material

```python
fd = open("Data3.txt","w+")
#fd.close()
```

how to Write using fd.writeLines

The issue is that you are opening the file in "w+" mode, which stands for "write and read." However, in this mode, when you write to the file, the file pointer is moved to the end of the file. So, when you try to read from the file using fd.read(), there is nothing left to read because the file pointer is already at the end after the write operation.

```python
list1 = ["This is Jalandhar \n","Time is 25 Jan 2025 \n","I will stay in room till 16Feb2025 \n"]
fd.write("Hello \n")
fd.writelines(list1)

fd.seek(0)  # Move the file pointer to the beginning

txt = fd.read()
print(txt)
fd.close()
```

### Difference BetWeen read and readline function

```python
file1 = open("Data3.txt","r+") 
  
print("Output of Read function is ")
print(file1.read())
print()
  
# seek(n) takes the file handle to the nth
# bite from the beginning.
file1.seek(0) 
  
print( "Output of Readline function is ")
print(file1.readline()) 
print()
  
file1.seek(0)
  
# To show difference between read and readline
print("Output of Read(9) function is ") 
print(file1.read(9))
print()
  
file1.seek(0)
  
print("Output of Readline(9) function is ") 
print(file1.readline(9))
  
file1.seek(0)
# readlines function
print("Output of Readlines function is ") 
print(file1.readlines()) 
print()
file1.close()
```
